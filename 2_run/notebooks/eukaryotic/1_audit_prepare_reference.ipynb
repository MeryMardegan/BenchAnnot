## Split eukaryotic audit reports

This notebook writes a GFF/GFFread audit and an independent NCBI reference audit per organism. The NCBI FAA remains the reference universe; GFFread is used only to audit extraction losses.

```python
from pathlib import Path

from benchannot.eukaryotic.audit_prepare_reference import (
run_pre_processing,
build_reference_audit
)
```

```python
PROJECT_ROOT = Path("../../..").resolve()
DATA_DIR = PROJECT_ROOT / "data"
GENOME_EUKARYOTE_DIR = DATA_DIR / "genome_eukaryote"
OUTPUT_DIR = PROJECT_ROOT / "2_run" / "output" / "eukaryotic"
AUDIT_DIR = OUTPUT_DIR / "audit"
GFFREAD_DIR = DATA_DIR / "origin" / "eukaryote_output_tools" / "gffread"
```

```python
ORGANISMS = [
    {
        "organism": "saccharomyces_cerevisiae",
        "original_gff": GENOME_EUKARYOTE_DIR / "saccharomyces_cerevisiae.gff",
        "reference_fasta": GENOME_EUKARYOTE_DIR / "saccharomyces_cerevisiae.faa",
        "filtered_gff": AUDIT_DIR / "gff_gffread" / "saccharomyces_cerevisiae" / "filtered.gff",
        "gffread_fasta": GFFREAD_DIR / "saccharomyces_cerevisiae_gffread.faa",
        "mitochondrial_prefix": "rna-Q",
        "output_directory": OUTPUT_DIR,
    },
    {
        "organism": "drosophila_melanogaster",
        "original_gff": GENOME_EUKARYOTE_DIR / "drosophila_melanogaster.gff",
        "reference_fasta": GENOME_EUKARYOTE_DIR / "drosophila_melanogaster.faa",
        "filtered_gff": AUDIT_DIR / "gff_gffread" / "drosophila_melanogaster" / "filtered.gff",
        "gffread_fasta": GFFREAD_DIR / "drosophila_melanogaster_gffread.faa",
        "mitochondrial_prefix": "rna-Dmel",
        "output_directory": OUTPUT_DIR,
    },
]
```

```python
for config in ORGANISMS:
    run_pre_processing(
        anno_sc=config["original_gff"],
        sample_id_sc=config["organism"],
        filtered_gff_sc=config["filtered_gff"],
    )
```

```python
results = {}

for config in ORGANISMS:
    name = config["organism"]

    results[name] = build_reference_audit(**config)
    result = results[name]
    print(
        f"[{name}] transcripts={len(result['transcript_reference'])} "
        f"loci={len(result['locus_reference'])} "
        f"gff_missing={len(result['post_gffread_missing'])}"
    )
```
