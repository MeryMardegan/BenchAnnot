# Preparation of annotation tool outputs

This notebook handles two independent sets of files:

1. **Original outputs** — files used in the published analysis and archived on Zenodo.
2. **Reproduced outputs** — files generated locally when rerunning the annotation tools.

Original files are never overwritten by reproduced files.

```python
import sys
import os
import requests
from pathlib import Path
from benchannot.io.download import (
download_file,
download_and_extract,
download_reference_genomes
)
```

```python
PROJECT_ROOT = Path("..").resolve()

DATA_GENOME_EUKARYOTE = PROJECT_ROOT / "data" / "genome_eukaryote"
DATA_GENOME_PROKARYOTE = PROJECT_ROOT / "data" / "genome_prokaryote"

DATA_ORIGIN = PROJECT_ROOT / "data" / "origin"
DATA_REPRODUCED = PROJECT_ROOT / "data" / "reproduced"

DATA_ORIGIN_EUKARYOTE = DATA_ORIGIN / "eukaryote_output_tools"
DATA_ORIGIN_PROKARYOTE = DATA_ORIGIN / "prokaryote_output_tools"

DATA_REPRODUCED_EUKARYOTE = DATA_REPRODUCED / "eukaryote_output_tools"
DATA_REPRODUCED_PROKARYOTE = DATA_REPRODUCED / "prokaryote_output_tools"
```

```python
# Annotation tools used for each organism group
eukaryote_subfolders = [
    "eggnog",
    "gffread",
    "interproscan",
    "kofamscan",
    "pannzer"
]

prokaryote_subfolders = [
    "bakta",
    "eggnog",
    "pgap",
    "prokka"
]

# Create required directories
for base_dir, subfolders in [
    (DATA_ORIGIN_EUKARYOTE, eukaryote_subfolders),
    (DATA_ORIGIN_PROKARYOTE, prokaryote_subfolders),
    (DATA_REPRODUCED_EUKARYOTE, eukaryote_subfolders),
    (DATA_REPRODUCED_PROKARYOTE, prokaryote_subfolders)
]:
    for folder in subfolders:
        folder_dir = base_dir / folder
        folder_dir.mkdir(parents=True, exist_ok=True)
        print(f"{folder_dir.relative_to(PROJECT_ROOT)} created.")
```
